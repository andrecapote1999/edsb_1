```python
import pandas as pd

SOURCE = "/kaggle/input/datasets/dilwong/flightprices/itineraries.csv"   # caminho do passo 4
OUTPUT = "/kaggle/working/itineraries_sample.csv"
FRACTION = 0.01                                         # 5% das linhas

first = True
total_in = total_out = 0
for chunk in pd.read_csv(SOURCE, chunksize=1_000_000, dtype=str):
    sample = chunk.sample(frac=FRACTION, random_state=42)
    sample.to_csv(OUTPUT, mode="w" if first else "a", header=first, index=False)
    first = False
    total_in += len(chunk)
    total_out += len(sample)
    print(f"lidas {total_in:,} | guardadas {total_out:,}")

print("Feito!")
```
